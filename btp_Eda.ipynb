```python
# Check for missing values
print("="*80)
print("MISSING VALUES ANALYSIS")
print("="*80)

missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

if missing.sum() == 0:
    print("✓ No missing values found in the dataset!")
else:
    missing_df = pd.DataFrame({
        'Missing_Count': missing,
        'Missing_Percentage': missing_pct
    })
    missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)
    display(missing_df)
```

```python
# ## 5. NOVA Classification Analysis

# %%
# NOVA class distribution
print("="*80)
print("NOVA CLASSIFICATION DISTRIBUTION")
print("="*80)
print("\nNOVA Classes:")
print("1 - Unprocessed or minimally processed foods")
print("2 - Processed culinary ingredients")
print("3 - Processed foods")
print("4 - Ultra-processed food products")

nova_dist = df['novaclass'].value_counts().sort_index()
nova_pct = (df['novaclass'].value_counts(normalize=True).sort_index() * 100)

nova_summary = pd.DataFrame({
    'Count': nova_dist,
    'Percentage': nova_pct
})
print("\n")
display(nova_summary)
```

```python
# Visualize NOVA class distribution
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Bar plot
nova_dist.plot(kind='bar', ax=axes[0], color='steelblue', edgecolor='black')
axes[0].set_title('NOVA Class Distribution (Count)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('NOVA Class', fontsize=12)
axes[0].set_ylabel('Count', fontsize=12)
axes[0].grid(axis='y', alpha=0.3)
for i, v in enumerate(nova_dist):
    axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold')

# Pie chart
axes[1].pie(nova_dist, labels=['Class 1', 'Class 2', 'Class 3', 'Class 4'],
            autopct='%1.1f%%', startangle=90, colors=['#2ecc71', '#f39c12', '#e74c3c', '#9b59b6'])
axes[1].set_title('NOVA Class Distribution (Percentage)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
## 6. Macronutrient Analysis

# %%
# Macronutrient statistics
macronutrients = ['Protein', 'Total Fat', 'Carbohydrate', 'Energy']

print("="*80)
print("MACRONUTRIENT ANALYSIS")
print("="*80)
display(df[macronutrients].describe())

# %%
# Distribution plots for macronutrients
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for i, nutrient in enumerate(macronutrients):
    axes[i].hist(df[nutrient], bins=50, color='skyblue', edgecolor='black', alpha=0.7)
    axes[i].axvline(df[nutrient].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df[nutrient].mean():.2f}')
    axes[i].axvline(df[nutrient].median(), color='green', linestyle='--', linewidth=2, label=f'Median: {df[nutrient].median():.2f}')
    axes[i].set_title(f'{nutrient} Distribution', fontsize=12, fontweight='bold')
    axes[i].set_xlabel(nutrient, fontsize=10)
    axes[i].set_ylabel('Frequency', fontsize=10)
    axes[i].legend()
    axes[i].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# %%
# Box plots for macronutrients
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

for i, nutrient in enumerate(macronutrients):
    axes[i].boxplot(df[nutrient], vert=True, patch_artist=True,
                    boxprops=dict(facecolor='lightblue', color='black'),
                    medianprops=dict(color='red', linewidth=2))
    axes[i].set_title(f'{nutrient}', fontsize=12, fontweight='bold')
    axes[i].set_ylabel('Value', fontsize=10)
    axes[i].grid(alpha=0.3)

plt.suptitle('Macronutrient Box Plots (Outlier Detection)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
## 7. Macronutrients by NOVA Class

# %%
# Macronutrient means by NOVA class
print("="*80)
print("MACRONUTRIENTS BY NOVA CLASS")
print("="*80)

for nova_class in sorted(df['novaclass'].unique()):
    print(f"\n{'='*40}")
    print(f"NOVA Class {nova_class}")
    print('='*40)
    nova_data = df[df['novaclass'] == nova_class][macronutrients]
    display(nova_data.describe())

# %%
# Grouped bar plot for macronutrients by NOVA class
nova_macro_means = df.groupby('novaclass')[macronutrients].mean()

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for i, nutrient in enumerate(macronutrients):
    nova_macro_means[nutrient].plot(kind='bar', ax=axes[i], color='coral', edgecolor='black')
    axes[i].set_title(f'{nutrient} by NOVA Class (Mean)', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('NOVA Class', fontsize=10)
    axes[i].set_ylabel(f'{nutrient} (g/kcal)', fontsize=10)
    axes[i].grid(axis='y', alpha=0.3)
    axes[i].set_xticklabels(axes[i].get_xticklabels(), rotation=0)

plt.tight_layout()
plt.show()

# %%
# Violin plots for macronutrients by NOVA class
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

for i, nutrient in enumerate(macronutrients):
    parts = axes[i].violinplot([df[df['novaclass']==j][nutrient].dropna()
                                for j in sorted(df['novaclass'].unique())],
                               positions=sorted(df['novaclass'].unique()),
                               showmeans=True, showmedians=True)
    axes[i].set_title(f'{nutrient} by NOVA Class', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('NOVA Class', fontsize=10)
    axes[i].set_ylabel(nutrient, fontsize=10)
    axes[i].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
## 8. Zero Values Analysis

# %%
# Analyze nutrients with high percentage of zeros
print("="*80)
print("ZERO VALUES ANALYSIS")
print("="*80)

zero_analysis = []
for col in df.columns:
    zero_count = (df[col] == 0).sum()
    zero_pct = (zero_count / len(df)) * 100
    zero_analysis.append({
        'Nutrient': col,
        'Zero_Count': zero_count,
        'Zero_Percentage': zero_pct,
        'Non_Zero_Count': len(df) - zero_count
    })

zero_df = pd.DataFrame(zero_analysis).sort_values('Zero_Percentage', ascending=False)
print("\nTop 20 Nutrients with Highest Zero Percentage:")
display(zero_df.head(20))

# %%
# Visualize nutrients with >50% zeros
high_zero_nutrients = zero_df[zero_df['Zero_Percentage'] > 50]

plt.figure(figsize=(12, 8))
plt.barh(high_zero_nutrients['Nutrient'][:20], high_zero_nutrients['Zero_Percentage'][:20],
         color='salmon', edgecolor='black')
plt.xlabel('Zero Percentage (%)', fontsize=12)
plt.ylabel('Nutrient', fontsize=12)
plt.title('Top 20 Nutrients with Highest Zero Percentage (>50%)', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
## 9. Correlation Analysis

# %%
# Correlation with Energy
print("="*80)
print("CORRELATION WITH ENERGY")
print("="*80)

energy_corr = df.corr()['Energy'].sort_values(ascending=False)
print("\nTop 15 nutrients correlated with Energy:")
display(pd.DataFrame(energy_corr.head(15)))

print("\nBottom 15 nutrients correlated with Energy:")
display(pd.DataFrame(energy_corr.tail(15)))

# %%
# Visualize top correlations with Energy
top_corr = energy_corr.drop('Energy').head(15)

plt.figure(figsize=(10, 8))
plt.barh(range(len(top_corr)), top_corr.values, color='teal', edgecolor='black')
plt.yticks(range(len(top_corr)), top_corr.index)
plt.xlabel('Correlation Coefficient', fontsize=12)
plt.ylabel('Nutrient', fontsize=12)
plt.title('Top 15 Nutrients Correlated with Energy', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# %%
# Correlation heatmap for macronutrients and key nutrients
key_nutrients = ['Protein', 'Total Fat', 'Carbohydrate', 'Energy', 'Water',
                 'Sugars, total', 'Fiber, total dietary', 'Sodium', 'Calcium',
                 'Iron', 'Vitamin C', 'Cholesterol']

plt.figure(figsize=(12, 10))
sns.heatmap(df[key_nutrients].corr(), annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Heatmap - Key Nutrients', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
## 10. Outlier Detection

# %%
# Outlier detection using IQR method
print("="*80)
print("OUTLIER DETECTION (IQR Method)")
print("="*80)

outliers_summary = []
key_nutrients_outlier = ['Protein', 'Total Fat', 'Carbohydrate', 'Energy', 'Sugars, total']

for nutrient in key_nutrients_outlier:
    Q1 = df[nutrient].quantile(0.25)
    Q3 = df[nutrient].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[nutrient] < lower_bound) | (df[nutrient] > upper_bound)]

    outliers_summary.append({
        'Nutrient': nutrient,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': len(outliers),
        'Outlier_Percentage': (len(outliers)/len(df)*100)
    })

outliers_df = pd.DataFrame(outliers_summary)
display(outliers_df)
```

```python
## 11. Special Compounds Analysis

# %%
# Analyze special compounds
print("="*80)
print("SPECIAL COMPOUNDS ANALYSIS")
print("="*80)

compounds = ['Caffeine', 'Theobromine', 'Alcohol', 'Cholesterol']
compounds_data = []

for compound in compounds:
    non_zero = df[df[compound] > 0]
    compounds_data.append({
        'Compound': compound,
        'Foods_With_Compound': len(non_zero),
        'Percentage_of_Foods': (len(non_zero)/len(df)*100),
        'Mean_Non_Zero': non_zero[compound].mean() if len(non_zero) > 0 else 0,
        'Median_Non_Zero': non_zero[compound].median() if len(non_zero) > 0 else 0,
        'Max_Value': df[compound].max(),
        'Min_Non_Zero': non_zero[compound].min() if len(non_zero) > 0 else 0
    })

compounds_df = pd.DataFrame(compounds_data)
display(compounds_df)

# %%
# Visualize special compounds presence
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for i, compound in enumerate(compounds):
    non_zero_data = df[df[compound] > 0][compound]
    if len(non_zero_data) > 0:
        axes[i].hist(non_zero_data, bins=30, color='mediumseagreen', edgecolor='black', alpha=0.7)
        axes[i].set_title(f'{compound} Distribution (Non-Zero Values)', fontsize=12, fontweight='bold')
        axes[i].set_xlabel(compound, fontsize=10)
        axes[i].set_ylabel('Frequency', fontsize=10)
        axes[i].grid(alpha=0.3)
        axes[i].text(0.7, 0.9, f'n = {len(non_zero_data)}',
                     transform=axes[i].transAxes, fontsize=10, verticalalignment='top',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    else:
        axes[i].text(0.5, 0.5, 'No non-zero values', ha='center', va='center',
                     transform=axes[i].transAxes, fontsize=12)

plt.tight_layout()
plt.show()
```

```python
## 12. Vitamin and Mineral Analysis

# %%
# Vitamin and mineral statistics
vitamins_minerals = ['Calcium', 'Iron', 'Magnesium', 'Phosphorus', 'Potassium',
                     'Sodium', 'Zinc', 'Vitamin A, RAE', 'Vitamin C',
                     'Vitamin D (D2 + D3)', 'Vitamin E (alpha-tocopherol)',
                     'Vitamin K (phylloquinone)', 'Vitamin B-12']

print("="*80)
print("VITAMIN AND MINERAL ANALYSIS")
print("="*80)
display(df[vitamins_minerals].describe())

# %%
# Box plots for vitamins and minerals
fig, axes = plt.subplots(3, 5, figsize=(20, 12))
axes = axes.ravel()

for i, nutrient in enumerate(vitamins_minerals):
    if i < len(axes):
        axes[i].boxplot(df[nutrient].dropna(), vert=True, patch_artist=True,
                        boxprops=dict(facecolor='lightcoral', color='black'),
                        medianprops=dict(color='blue', linewidth=2))
        axes[i].set_title(f'{nutrient}', fontsize=10, fontweight='bold')
        axes[i].set_ylabel('Value', fontsize=9)
        axes[i].grid(alpha=0.3)

# Hide unused subplots
for i in range(len(vitamins_minerals), len(axes)):
    axes[i].axis('off')

plt.suptitle('Vitamin and Mineral Box Plots', fontsize=14, fontweight='bold', y=1.0)
plt.tight_layout()
plt.show()
```

```python
## 13. Fatty Acid Composition

# %%
# Fatty acid analysis
fatty_acids = ['Fatty acids, total saturated', 'Fatty acids, total monounsaturated',
               'Fatty acids, total polyunsaturated']

print("="*80)
print("FATTY ACID COMPOSITION ANALYSIS")
print("="*80)
display(df[fatty_acids].describe())

# %%
# Fatty acid composition visualization
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, fa in enumerate(fatty_acids):
    axes[i].hist(df[fa], bins=50, color='plum', edgecolor='black', alpha=0.7)
    axes[i].axvline(df[fa].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df[fa].mean():.2f}')
    axes[i].set_title(fa, fontsize=11, fontweight='bold')
    axes[i].set_xlabel('Value (g)', fontsize=10)
    axes[i].set_ylabel('Frequency', fontsize=10)
    axes[i].legend()
    axes[i].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# %%
# Fatty acid ratio by NOVA class
fa_by_nova = df.groupby('novaclass')[fatty_acids].mean()

fig, ax = plt.subplots(figsize=(12, 6))
fa_by_nova.plot(kind='bar', ax=ax, color=['#e74c3c', '#3498db', '#2ecc71'],
                edgecolor='black', width=0.8)
ax.set_title('Fatty Acid Composition by NOVA Class (Mean)', fontsize=14, fontweight='bold')
ax.set_xlabel('NOVA Class', fontsize=12)
ax.set_ylabel('Value (g)', fontsize=12)
ax.legend(title='Fatty Acid Type', fontsize=10)
ax.grid(axis='y', alpha=0.3)
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.show()
```
