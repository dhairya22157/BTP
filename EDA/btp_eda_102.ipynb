```python
print("="*80)
print("DATASET OVERVIEW")
print("="*80)
print(f"\nMemory Usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
print("\nData Types:")
print(df.dtypes.value_counts())

print("\n" + "="*80)
print("NEW FLAVONOID/PHYTOCHEMICAL FEATURES")
print("="*80)
flavonoid_keywords = ['flavon', 'catechin', 'anthocyan', 'isoflavone', 'theaflavin']
new_nutrients = [col for col in df.columns if any(k in col.lower() for k in flavonoid_keywords)]
print(f"\nTotal new phytochemical features: {len(new_nutrients)}")
for i, nutrient in enumerate(new_nutrients[:15], 1):
    print(f"{i}. {nutrient}")

print("\n\nFirst 10 Rows:")
display(df.head(10))
```

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

if missing.sum() == 0:
    print("✓ No missing values found! Excellent data quality.")
else:
    missing_df = pd.DataFrame({
        'Missing_Count': missing[missing>0],
        'Missing_Percentage': missing_pct[missing_pct>0]
    }).sort_values('Missing_Count', ascending=False)
    display(missing_df)
```

```python
print("STATISTICAL SUMMARY")
print("="*80)
display(df.describe().T)
```

```python
print("="*80)
print("NOVA CLASSIFICATION")
print("="*80)
print("\nNOVA Classes:")
print("1 - Unprocessed/minimally processed (highest natural phytochemicals)")
print("2 - Processed culinary ingredients")
print("3 - Processed foods")
print("4 - Ultra-processed (lowest natural compounds)\n")

nova_dist = df['novaclass'].value_counts().sort_index()
nova_pct = (nova_dist / len(df) * 100)

nova_summary = pd.DataFrame({
    'Count': nova_dist,
    'Percentage': nova_pct.round(2)
})
display(nova_summary)
```

```python
# CELL 8: NOVA Visualization
# ============================================================================
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

nova_dist.plot(kind='bar', ax=axes[0], color='steelblue', edgecolor='black')
axes[0].set_title('NOVA Class Distribution', fontsize=14, fontweight='bold')
axes[0].set_xlabel('NOVA Class', fontsize=12)
axes[0].set_ylabel('Count', fontsize=12)
axes[0].grid(axis='y', alpha=0.3)
for i, v in enumerate(nova_dist):
    axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold')

axes[1].pie(nova_dist, labels=['Class 1', 'Class 2', 'Class 3', 'Class 4'],
            autopct='%1.1f%%', startangle=90,
            colors=['#2ecc71', '#f39c12', '#e74c3c', '#9b59b6'])
axes[1].set_title('NOVA Distribution (%)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# CELL 9: Macronutrient Analysis
# ============================================================================
macronutrients = ['Protein', 'Total Fat', 'Carbohydrate', 'Energy']

print("="*80)
print("MACRONUTRIENT STATISTICS")
print("="*80)
display(df[macronutrients].describe())
```

```python
# CELL 10: Macronutrient Distributions
# ============================================================================
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for i, nutrient in enumerate(macronutrients):
    axes[i].hist(df[nutrient], bins=50, color='skyblue', edgecolor='black', alpha=0.7)
    axes[i].axvline(df[nutrient].mean(), color='red', linestyle='--', linewidth=2,
                    label=f'Mean: {df[nutrient].mean():.2f}')
    axes[i].axvline(df[nutrient].median(), color='green', linestyle='--', linewidth=2,
                    label=f'Median: {df[nutrient].median():.2f}')
    axes[i].set_title(f'{nutrient} Distribution', fontsize=12, fontweight='bold')
    axes[i].set_xlabel(nutrient, fontsize=10)
    axes[i].set_ylabel('Frequency', fontsize=10)
    axes[i].legend()
    axes[i].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# CELL 11: Zero Values Analysis
# ============================================================================
print("="*80)
print("ZERO VALUES ANALYSIS")
print("="*80)

zero_analysis = []
for col in df.columns:
    zero_count = (df[col] == 0).sum()
    zero_pct = (zero_count / len(df)) * 100
    zero_analysis.append({
        'Nutrient': col,
        'Zero_Count': zero_count,
        'Zero_Percentage': zero_pct,
        'Non_Zero_Count': len(df) - zero_count
    })

zero_df = pd.DataFrame(zero_analysis).sort_values('Zero_Percentage', ascending=False)
print(f"\nNutrients with >50% zeros: {len(zero_df[zero_df['Zero_Percentage'] > 50])}")
print(f"Nutrients with >90% zeros: {len(zero_df[zero_df['Zero_Percentage'] > 90])}")
print("\nTop 25 Nutrients with Highest Zero Percentage:")
display(zero_df.head(25))
```

```python
# CELL 12: Zero Values Visualization
# ============================================================================
high_zero = zero_df[zero_df['Zero_Percentage'] > 50].head(25)

plt.figure(figsize=(12, 10))
plt.barh(high_zero['Nutrient'], high_zero['Zero_Percentage'],
         color='salmon', edgecolor='black')
plt.xlabel('Zero Percentage (%)', fontsize=12)
plt.ylabel('Nutrient', fontsize=12)
plt.title('Top 25 Nutrients with >50% Zero Values', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# CELL 13: FLAVONOID ANALYSIS (NEW!)
# ============================================================================
print("="*80)
print("FLAVONOID & PHYTOCHEMICAL ANALYSIS")
print("="*80)
print("\nThese bioactive compounds are KEY for:")
print("- Antioxidant properties")
print("- Health benefits")
print("- Flavor profiles")
print("- Phytochemical-based food pairing\n")

flavonoid_cols = ['Total flavonoids', 'Total anthocyanidins',
                  'Total catechins (monomeric flavan-3-ols only)',
                  'Total flavan-3-ols', 'Total flavanones', 'Total flavones',
                  'Total flavonols', 'Total isoflavones']

flavonoid_summary = []
for col in flavonoid_cols:
    non_zero = df[df[col] > 0]
    zero_pct = ((df[col] == 0).sum() / len(df)) * 100
    flavonoid_summary.append({
        'Compound': col.replace('Total ', ''),
        'Foods_Count': len(non_zero),
        'Foods_Percentage': f"{(len(non_zero)/len(df)*100):.2f}%",
        'Mean_Non_Zero': f"{non_zero[col].mean():.2f}" if len(non_zero) > 0 else '0',
        'Median_Non_Zero': f"{non_zero[col].median():.2f}" if len(non_zero) > 0 else '0',
        'Max_Value': f"{df[col].max():.2f}"
    })

flavonoid_df = pd.DataFrame(flavonoid_summary)
display(flavonoid_df)
```

```python
# CELL 14: Flavonoid Presence Visualization
# ============================================================================
flavonoid_presence = [len(df[df[col] > 0]) for col in flavonoid_cols]
flavonoid_names = [col.replace('Total ', '') for col in flavonoid_cols]

plt.figure(figsize=(12, 6))
bars = plt.bar(range(len(flavonoid_names)), flavonoid_presence,
               color='mediumseagreen', edgecolor='black', alpha=0.8)
plt.xticks(range(len(flavonoid_names)), flavonoid_names, rotation=45, ha='right')
plt.xlabel('Flavonoid Type', fontsize=12)
plt.ylabel('Number of Foods', fontsize=12)
plt.title('Flavonoid Presence Across Food Items', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)

# Add value labels
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(height)}\n({height/len(df)*100:.1f}%)',
             ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# CELL 15: Total Flavonoids Distribution
# ============================================================================
flavonoid_foods = df[df['Total flavonoids'] > 0]['Total flavonoids']

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Histogram
axes[0].hist(flavonoid_foods, bins=50, color='purple', edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Total Flavonoids (mg)', fontsize=12)
axes[0].set_ylabel('Frequency', fontsize=12)
axes[0].set_title('Flavonoid Distribution (Non-Zero)', fontsize=12, fontweight='bold')
axes[0].grid(alpha=0.3)

# Box plot
axes[1].boxplot(flavonoid_foods, vert=True, patch_artist=True,
                boxprops=dict(facecolor='plum', color='black'),
                medianprops=dict(color='red', linewidth=2))
axes[1].set_ylabel('Total Flavonoids (mg)', fontsize=12)
axes[1].set_title('Flavonoid Box Plot', fontsize=12, fontweight='bold')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f'Foods with flavonoids: {len(flavonoid_foods)} ({len(flavonoid_foods)/len(df)*100:.1f}%)')
print(f'Mean: {flavonoid_foods.mean():.2f} mg')
print(f'Median: {flavonoid_foods.median():.2f} mg')
print(f'Max: {flavonoid_foods.max():.2f} mg')
```

```python
# CELL 16: Flavonoids by NOVA Class
# ============================================================================
print("="*80)
print("FLAVONOID CONTENT BY NOVA CLASS")
print("="*80)
print("\nExpectation: Class 1 (unprocessed) should have highest flavonoid content")
print("Class 4 (ultra-processed) should have minimal natural flavonoids\n")

for nova in sorted(df['novaclass'].unique()):
    subset = df[df['novaclass'] == nova]
    non_zero = subset[subset['Total flavonoids'] > 0]
    print(f"\nNOVA Class {nova}:")
    print(f"  Foods with flavonoids: {len(non_zero)}/{len(subset)} ({len(non_zero)/len(subset)*100:.1f}%)")
    if len(non_zero) > 0:
        print(f"  Mean: {non_zero['Total flavonoids'].mean():.2f} mg")
        print(f"  Median: {non_zero['Total flavonoids'].median():.2f} mg")
```

```python
# CELL 17: Correlation with Energy
# ============================================================================
print("="*80)
print("CORRELATION WITH ENERGY")
print("="*80)

energy_corr = df.corr()['Energy'].sort_values(ascending=False)
print("\nTop 20 nutrients correlated with Energy:")
display(pd.DataFrame(energy_corr.head(20)))
```

```python
# CELL 18: Correlation Heatmap (Including Flavonoids)
# ============================================================================
key_nutrients = ['Protein', 'Total Fat', 'Carbohydrate', 'Energy', 'Water',
                 'Fiber, total dietary', 'Total flavonoids', 'Total anthocyanidins',
                 'Total catechins (monomeric flavan-3-ols only)', 'Total flavonols']

plt.figure(figsize=(12, 10))
sns.heatmap(df[key_nutrients].corr(), annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Heatmap - Macros & Flavonoids', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# CELL 19: Outlier Detection
# ============================================================================
print("="*80)
print("OUTLIER DETECTION (IQR Method)")
print("="*80)

outliers_summary = []
key_nutrients_outlier = ['Protein', 'Total Fat', 'Carbohydrate', 'Energy',
                         'Sugars, total', 'Total flavonoids']

for nutrient in key_nutrients_outlier:
    Q1 = df[nutrient].quantile(0.25)
    Q3 = df[nutrient].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[nutrient] < lower_bound) | (df[nutrient] > upper_bound)]

    outliers_summary.append({
        'Nutrient': nutrient,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': len(outliers),
        'Outlier_Percentage': (len(outliers)/len(df)*100)
    })

outliers_df = pd.DataFrame(outliers_summary)
display(outliers_df)
```
